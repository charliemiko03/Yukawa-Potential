```python
import numpy as np
import matplotlib.pyplot as plt

# =========================
# Parameters
# =========================
g = 1.5        # strength (init 1.5)
lam = 0.5      # screening parameter (init 0.5)

# Avoid r = 0 (singularity)
r = np.linspace(0.01, 10, 1000)

# Yukawa potential
V = -g * np.exp(-lam * r) / r

# =========================
# Plot
# =========================
plt.figure(figsize=(8,5))
plt.plot(r, V)
plt.xlabel('r')
plt.ylabel('V(r)')
plt.title('Yukawa Potential')
plt.axhline(0, linewidth=0.8)
plt.grid(True)
plt.show()
```

```python
Vc = -g / r  # Coulomb potential

plt.figure(figsize=(8,5))
plt.plot(r, V, label='Yukawa')
plt.plot(r, Vc, linestyle='--', label='Coulomb')
plt.xlabel('r')
plt.ylabel('V(r)')
plt.title('Yukawa vs Coulomb Potential')
plt.legend()
plt.grid(True)
plt.show()
```
